```python
df['htm_weekday'] = df['htm_weekday'].fillna(df['htm_weekday'].median())
df['htm_weekend'] = df['htm_weekend'].fillna(df['htm_weekend'].median())
df['description'] = df['description'].fillna('')
df.isnull().sum()
```

```python
indonesian_stopwords = ["yang", "di", "dan", "adalah", "ini", "itu", "untuk", "dengan", "pada", "ke", "dari"]
def clean_text(text):
    if not isinstance(text, str): return ""
    text = text.lower() # Kecilkan huruf
    text = re.sub(r'[^a-z\s]', '', text) # Hapus angka dan simbol
    words = text.split()
    words = [w for w in words if w not in indonesian_stopwords and len(w) > 1]
    return " ".join(words)

df['description_clean'] = df['description'].apply(clean_text)
df_clean = df.copy()
df_clean.head()
df_clean.info()
```

```python
def kategori_harga(x):
    if x < 10000:
        return "Murah"
    elif x <= 50000:
        return "Menengah"
    else:
        return "Mahal"

df_clean['kategori_harga_weekday'] = df_clean['htm_weekday'].apply(kategori_harga)
df_clean['kategori_harga_weekend'] = df_clean['htm_weekend'].apply(kategori_harga)
```

```python
def rating_segment(x):
    if x >= 4.5:
        return "Sangat Baik"
    elif x >= 4:
        return "Baik"
    elif x >= 3:
        return "Cukup"
    else:
        return "Perlu Perbaikan"

df_clean['rating_segment'] = df_clean['vote_average'].apply(rating_segment)
```
